```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ All libraries imported successfully!")
```

```python
df=pd.read_csv("../Data/Cleaned data/imdb_cleaned.csv")
```

```python
#1. How are IMDb ratings distributed across all movies?
plt.figure(figsize= (6,4))
plt.hist(df["IMDB_Rating"], bins= 20, edgecolor= "black")
plt.grid(alpha=0.3)
plt.title("Distribution of IMDb Ratings")
plt.xlabel("Ratings")
plt.ylabel("Number of movies")
plt.show()
```

```python
#2. Which genres have the highest average IMDb ratings?
avg_rating = df.groupby("Genre")["IMDB_Rating"].mean()
plt.figure(figsize=(6,4))
plt.bar(avg_rating .index,avg_rating .values, edgecolor="black" )
plt.show()
```

```python
#2. Which genres have the highest average IMDb ratings?
avg_rating = (df.groupby("Genre")["IMDB_Rating"].mean().sort_values(ascending=False).head(10))

plt.figure(figsize=(10,5))
plt.bar(avg_rating.index, avg_rating.values, edgecolor="black")

plt.title("Top 10 Genre Combinations by Average IMDb Rating")
plt.xlabel("Genre")
plt.ylabel("Average IMDb Rating")
plt.xticks(rotation=45, ha="right")

plt.show()
```

```python
#3.How has the number of movies changed over the years?
count= df["Released_Year"].value_counts().sort_index(ascending=True)
plt.figure(figsize=(10,6))
plt.plot(count.index, count.values)
plt.title("Number of Movies Released Over the Years")
plt.xlabel("Released Year")
plt.ylabel("Number of Movies")
plt.xticks(count.index[::10], rotation=45)
plt.show()
```

```python
#4. relationship between Gross Revenue and IMDb Rating
plt.figure(figsize=(10,6))
plt.scatter(df["Gross"], df["IMDB_Rating"])
plt.title("Gross Revenue vs IMDb Rating")
plt.xlabel("Gross Revenue")
plt.ylabel("IMDb Rating")
plt.show()
```

```python
#5. Which directors have directed the most movies in this dataset?
director_count = df["Director"].value_counts().head(10)
plt.figure(figsize=(10,6))
plt.bar(director_count.index, director_count.values, edgecolor="black")
plt.title("Top 10 Directors by Number of Movies")
plt.xlabel("Director")
plt.ylabel("Number of Movies")

plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
#6. What is the percentage distribution of movie certificates (U, UA, A, PG, etc.) in the dataset?
certificate = df["Certificate"].value_counts()
plt.figure(figsize=(7,7))

plt.pie(certificate.values,labels=certificate.index,autopct="%1.1f%%",startangle=90)

plt.title("Distribution of Movie Certificates")
#to create a legend at side
plt.legend(certificate.index,title="Certificate",loc="center left",bbox_to_anchor=(1, 0.5))

plt.show()
```
