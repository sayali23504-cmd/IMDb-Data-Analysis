```python
import pandas as pd
import numpy as np


print("✅ All libraries imported successfully!")
```

```python
df=pd.read_csv("../Data/Cleaned data/imdb_cleaned.csv")
```

```python
#scenario 1:
#Which are the Top 20 movies based on IMDb rating?
top_20= df.sort_values(by="IMDB_Rating", ascending= False)
print(top_20[["Series_Title","IMDB_Rating"]].head(20))
```

```python
#scenario 2:
# What is the average runtime of all movies?
print("average runtime of all movies:", df["Runtime"].mean())
#Average runtime by genre
avg_runtime= df.groupby("Genre")["Runtime"].mean()
print("Average runtime by genre:", avg_runtime)
```

```python
#scenario 3:
# How many movies are longer than 180 minutes?
n_movies= df[df["Runtime"]>180]
print(n_movies.shape[0])
```

```python
#scenario 4:
# Which genre has the highest average IMDb rating?
avg_rating = df.groupby("Genre")["IMDB_Rating"].mean()
print("Genre:", avg_rating.idxmax())
print("Rating:", avg_rating.max())
```
